# Queue pakai list

```python
# Program Python untuk mendemonstrasikan implementasi queue menggunakan list
queue = []

# Fungsi append() digunakan untuk menambahkan elemen ke dalam queue (enqueue)
queue.append('a')
queue.append('i')
queue.append('u')
queue.append('e')
queue.append('o')

print("Queue awal")
print(queue)

# Fungsi pop() digunakan untuk menghapus elemen dari queue dengan urutan FIFO
print("\nElemen yang dikeluarkan dari queue:")
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

print("\nQueue setelah beberapa elemen dihapus")
print(queue)

# Jika baris print(queue.pop(0)) diaktifkan lagi,
# maka akan terjadi IndexError karena queue sudah kosong
```

# Queue menggunakan collections.deque

```python
# Program Python untuk mendemonstrasikan implementasi queue menggunakan collections.deque
from collections import deque

queue = deque()

# Fungsi append() digunakan untuk menambahkan elemen ke dalam queue
queue.append('a')
queue.append('i')
queue.append('u')
queue.append('e')
queue.append('o')

print("Queue awal:")
print(queue)

# Fungsi popleft() digunakan untuk menghapus elemen pertama dari queue (FIFO)
print("\nElemen yang dikeluarkan dari queue")
print(queue.popleft())
print(queue.popleft())
print(queue.popleft())

print("\nQueue setelah beberapa elemen dihapus")
print(queue)

# Jika baris print(queue.popleft()) diaktifkan lagi,
# maka akan terjadi IndexError karena queue sudah kosong
```

# Queue menggunakan modul queue

```python
# Program Python untuk mendemonstrasikan implementasi queue menggunakan modul queue
from queue import Queue

# Membuat queue dengan kapasitas maksimal 5
qu = Queue(maxsize=5)

# qsize() digunakan untuk menampilkan jumlah elemen dalam queue
print(qu.qsize())

# Fungsi put() digunakan untuk menambahkan elemen ke dalam queue
qu.put('a')
qu.put('i')
qu.put('u')
qu.put('e')
qu.put('o')

print("Full:", qu.full())
print("Size:", qu.qsize())

# Fungsi get() digunakan untuk mengambil atau menghapus elemen dari queue
print("\nElemen yang dikeluarkan dari queue")
print(qu.get())
print(qu.get())
print(qu.get())

print("\nEmpty:", qu.empty())
print("Full:", qu.full())
```

# Implementasi Queue menggunakan Class Linkedlist

```python
class Queue:
    def __init__(self):
        self.queue = list()

    # Fungsi untuk menambahkan data ke dalam queue
    def addtoqu(self, dataval):
        # Data hanya ditambahkan jika belum ada di dalam queue
        if dataval not in self.queue:
            self.queue.insert(0, dataval)
            return True
        return False

    # Fungsi pop untuk menghapus elemen dari queue
    def removefromq(self):
        if len(self.queue) > 0:
            return self.queue.pop()
        return ("Tidak ada elemen di dalam queue!")

    def size(self):
        return len(self.queue)


TheQueue = Queue()
print(TheQueue.addtoqu("Jan"))
print(TheQueue.addtoqu("Feb"))
print(TheQueue.addtoqu("March"))
print(TheQueue.addtoqu("April"))

print(TheQueue.size())

print(TheQueue.removefromq())
print(TheQueue.removefromq())

print(TheQueue.size())
```
