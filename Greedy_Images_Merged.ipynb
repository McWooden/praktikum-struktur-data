# Coin Charge

```python
def coin_change_greedy(amount, coins):
    coins.sort(reverse=True)
    result = []

    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)

    return result

# Contoh penggunaan
amount = 57
coins = [25, 10, 5, 1]
change = coin_change_greedy(amount, coins)
print("Koin yang digunakan:", change)
```

# Fractional Knapsack

```python
def fractional_knapsack(items, capacity):
    # items = list of tuples (value, weight)
    items = sorted(items, key=lambda x: x[0]/x[1], reverse=True)
    total_value = 0.0
    for value, weight in items:
        if capacity >= weight:
            capacity -= weight
            total_value += value
        else:
            total_value += value * (capacity / weight)
            break
    return total_value
# Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)]  # (nilai, berat)
capacity = 50
max_value = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
```

# Minimum Spanning Tree

```python
import heapq

def prim_mst(graph, start):
    visited = set()
    min_heap = [(0, start)]
    total_weight = 0
    while min_heap:
        weight, node = heapq.heappop(min_heap)
        if node in visited:
            continue
        visited.add(node)
        total_weight += weight
        for neighbor, edge_weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor))
    return total_weight
# Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
print("Total bobot MST (Prim):", prim_mst(graph, 'A'))
```

# Kruskal's Algorithm

```python
def kruskal_mst(edges, n_nodes):
    parent = {i: i for i in range(n_nodes)}
    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x
    def union(x, y):
        root_x = find(x)
        root_y = find(y)
        if root_x != root_y:
            parent[root_y] = root_x
            return True
        return False
    edges.sort(key=lambda x: x[2])
    total_weight = 0
    for u, v, weight in edges:
        if union(u, v):
            total_weight += weight
    return total_weight
# edges: (node1, node2, weight)
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print("Total bobot MST (Kruskal):", kruskal_mst(edges, 4))
```

# Latihan Personal

```python
import networkx as nx
import matplotlib.pyplot as plt

def visualize_mst(graph_dict, mst_edges):
    G = nx.Graph()
    for node, neighbors in graph_dict.items():
        for neighbor, weight in neighbors:
            G.add_edge(node, neighbor, weight=weight)
    pos = nx.spring_layout(G)
    edge_labels = nx.get_edge_attributes(G, 'weight')
    plt.figure(figsize=(8, 6))
    nx.draw(G, pos, with_labels=True, node_color='lightblue',
            node_size=2000, font_weight='bold')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
    nx.draw_networkx_edges(G, pos, edgelist=mst_edges,
                           edge_color='red', width=3)
    plt.title("Visualisasi MST")
    plt.axis('off')
    plt.show()
# Graf yang sama, MST hasil Prim
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
# MST edges hasil manual (sesuaikan dengan output Prim/Kruskal kamu)
mst_edges = [('A', 'B'), ('B', 'C'), ('B', 'D')]
visualize_mst(graph, mst_edges)
```
