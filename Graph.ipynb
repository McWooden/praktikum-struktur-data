# Membuat graf tak berarah menggunakan adjacency list

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}


# Menampilkan graf
def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")


print_graph(graph)
```

# Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])


    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# Contoh pemanggilan
bfs(graph, 'A')
```

# Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()


    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)


# Contoh pemanggilan
dfs(graph, 'A')
```

# Latihan personal

```python
def count_neighbors(graph):
    result = {}
    for node in graph:
        result[node] = len(graph[node])  # ini yang menggantikan pass
    return result

print(count_neighbors(graph)) # variabel graph ada di atas
# Output: {'A': 2, 'B': 3, 'C': 2, 'D': 1, 'E': 2, 'F': 2}
```

# Tugas

```python
# 1. Graf baru minimal 5 simpul & 7 sisi
graph = {
    'Semarang':   ['Solo', 'Demak', 'Kendal'],
    'Solo':       ['Semarang', 'Yogyakarta', 'Klaten'],
    'Demak':      ['Semarang', 'Kudus'],
    'Kendal':     ['Semarang'],
    'Yogyakarta': ['Solo'],
    'Klaten':     ['Solo'],
    'Kudus':      ['Demak']
}

print_graph(graph)

# 2. BFS dan DFS dari simpul awal
bfs(graph, 'Semarang') # Semarang Solo Demak Kendal Yogyakarta Klaten Kudus

dfs(graph, 'Semarang') # Semarang Solo Yogyakarta Klaten Demak Kudus Kendal

# 3 BFS yang mengembalikan list, bukan hanya print
def bfs_list(graph, start):
    visited = set()
    queue = deque([start])
    order = []

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            order.append(vertex)
            visited.add(vertex)
            queue.extend([n for n in graph[vertex] if n not in visited])

    return order

print(bfs_list(graph, 'Semarang')) # ['Semarang', 'Solo', 'Demak', 'Kendal', 'Yogyakarta', 'Klaten', 'Kudus']

# 4 find_path() menggunakan DFS
def find_path(graph, start, end, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)

    if start == end:
        return [start]

    for neighbor in graph[start]:
        if neighbor not in visited:
            path = find_path(graph, neighbor, end, visited)
            if path:
                return [start] + path

    return None

print(find_path(graph, 'Kudus', 'Yogyakarta')) # ['Kudus', 'Demak', 'Semarang', 'Solo', 'Yogyakarta']

# 5. is_connected() menggunakan DFS
def is_connected(graph):
    if not graph:
        return True

    def dfs_visit(node, visited):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_visit(neighbor, visited)

    start = next(iter(graph))
    visited = set()
    dfs_visit(start, visited)

    return len(visited) == len(graph)

print(is_connected(graph))  # True

graph_putus = {
    'A': ['B'], 'B': ['A'],
    'C': ['D'], 'D': ['C']  # terisolasi dari A-B
}
print(is_connected(graph_putus))  # False
```
